```python
import matplotlib.pyplot as plt
import mesa_reader as mr
import numpy as np
```

```python
log = mr.MesaLogDir('LOGS')

p1 = log.profile_data(model_number=1)
p2 = log.profile_data(model_number=50)
p3 = log.profile_data(model_number=100)
p4 = log.profile_data(model_number=150)
p5 = log.profile_data(model_number=200)
p6 = log.profile_data(model_number=250)
p7 = log.profile_data(model_number=-1)
print(p2,p3,p4)
print(p4,p5,p6,p7)
```

```python
cmap = plt.get_cmap('plasma')
colors = cmap(np.linspace(0, 1, 4)) 
# H
plt.plot(p4.mass, p4.x_mass_fraction_H, label=r't = 1.42 x $10^7$ yr (ZAMS)', linewidth=2,color=colors[0])
plt.plot(p5.mass, p5.x_mass_fraction_H, label=r't = 2.75 x $10^9$ yr', linewidth=2,color=colors[1])
plt.plot(p6.mass, p6.x_mass_fraction_H, label=r't = 7.88 x $10^9$ yr', linewidth=2,color=colors[2])
plt.plot(p7.mass, p7.x_mass_fraction_H, label=r't = 8.76 x $10^9$ yr (TAMS)', linewidth=2,color=colors[3])

plt.legend()

plt.xlabel(r'Mass Coordinate ($M_r / M_\odot$)', fontsize=12)
plt.ylabel('Hydrogen Mass Fraction ($X$)', fontsize=12)
plt.title('Hydrogen Abundance as a function of mass profile', fontsize=14)
plt.xlim(0, 0.5)
plt.grid()

plt.show()
```

```python
plt.plot(p7.mass, p7.pp, color=colors[3], linewidth=2, label = r't = 8.76 x $10^9$ yr (TAMS)')
plt.plot(p6.mass, p6.pp, color=colors[2], linewidth=2,label=r't = 7.88 x $10^9$ yr')
plt.plot(p5.mass, p5.pp, color=colors[1], linewidth=2,label = r't = 2.75 x $10^9$ yr')
plt.plot(p4.mass, p4.pp, color=colors[0], linewidth=2,label = r't = 1.42 x $10^7$ yr (ZAMS)')
plt.xlabel(r'Mass Coordinate ($M_r / M_\odot$)', fontsize=12)
plt.ylabel(r'Nuclear Energy Generation Rate, pp', fontsize=12)
plt.title('Energy Generation Profile during 4 stages of main sequence', fontsize=14)


plt.xlim(0, 0.5) 
plt.grid()
plt.legend()
```

```python
fig1, (ax1,ax2) = plt.subplots(1,2,figsize=(8, 4))

ax1.set_xlabel(r'Mass Coordinate ($M_r / M_\odot$)', fontsize=12)
ax1.set_ylabel(r'Temperature ($10^7$ K)', fontsize=12)
ax1.plot(p7.mass,p7.T / 1e7, color=colors[3], linewidth=2, label=r't = 8.76 x $10^9$ yr (TAMS)')
ax1.plot(p6.mass,p6.T / 1e7, color=colors[2], linewidth=2, label=r't = 7.88 x $10^9$ yr')
ax1.plot(p5.mass,p5.T / 1e7, color=colors[1], linewidth=2, label=r't = 2.75 x $10^9$ yr')
ax1.plot(p4.mass,p4.T / 1e7, color=colors[0], linewidth=2, label=r't = 1.42 x $10^7$ yr (ZAMS)')
ax1.legend()
ax1.set_title('mass profile vs. temperature')
ax1.grid()

ax2.set_xlabel(r'Mass Coordinate ($M_r / M_\odot$)', fontsize=12)
ax2.set_ylabel(r'Density (g/cm$^3$)', fontsize=12)
ax2.plot(p7.mass, 10**p7.logRho, color=colors[3], linewidth=2, linestyle='--', label=r't = 8.76 x $10^9$ yr (TAMS)')
ax2.plot(p6.mass, 10**p6.logRho, color=colors[2], linewidth=2, linestyle='--', label=r't = 7.88 x $10^9$ yr')
ax2.plot(p5.mass, 10**p5.logRho, color=colors[1], linewidth=2, linestyle='--', label=r't = 2.75 x $10^9$ yr')
ax2.plot(p4.mass, 10**p4.logRho, color=colors[0], linewidth=2, linestyle='--', label=r't = 1.42 x $10^7$ yr (ZAMS)')
ax2.set_title('mass profile vs. density')
ax2.legend()
ax2.grid()

fig1.tight_layout()
```

```python
plt.plot(10**(p7.logR),p7.y_mass_fraction_He,color=colors[3],label=r't = 8.76 x $10^9$ yr (TAMS)')
plt.plot(10**(p6.logR),p6.y_mass_fraction_He,color=colors[2],label=r't = 7.88 x $10^9$ yr')
plt.plot(10**(p5.logR),p5.y_mass_fraction_He,color=colors[1],label=r't = 2.75 x $10^9$ yr')
plt.plot(10**(p4.logR),p4.y_mass_fraction_He,color=colors[0],label = r't = 1.42 x $10^7$ yr (ZAMS)')
plt.axvline(0.03,color='red')
plt.ylabel('He mass fraction',size=14)
plt.xlabel(r'Radius of inside the star ($R_\odot$)',size=14)
plt.title('He core radius at different time during Main Sequence',size=16)
plt.xlim(0,0.3)
plt.ylim(0,1)
plt.legend()
plt.grid()
plt.show()
```
